# claude-plugins

> One-command Claude Code dev environment setup — hooks, MCP servers, and skills for the AnswerDotAI ecosystem

## Installation

```bash
pip install claude-plugins
# or with uv:
uv add claude-plugins
```

Then run setup in your project directory:

```bash
claude-plugins setup
# or:
uv run claude-plugins setup
```

This single command:
- Installs all dev packages (fastHTML, litesearch, Lisette, exhash, safecmd, safepyrun, codesigs, nbdev)
- Wires up Claude Code hooks (safecmd, exhash, code indexer)
- Registers MCP servers (exhash, safecmd, safepyrun)
- Installs skills for all packages
- Runs `nbdev-install-hooks` for git/Jupyter integration
- Builds initial code index

## What's Included

### Hooks

| Hook | Event | What it does |
|------|-------|--------------|
| safecmd | PreToolUse (Bash) | Validates every shell command against an allowlist before running |
| exhash | PostToolUse (Edit/Write) | Logs hash-addressed snapshot of every file edit |
| indexer | SessionStart | Indexes codebase with codesigs + litesearch |

### MCP Servers

| Server | Tools exposed |
|--------|---------------|
| exhash | `lnhashview`, `exhash_edit` |
| safecmd | `validate_command`, `run_safe` |
| safepyrun | `run_python`, `allow_function` |

### Skills

`/exhash`, `/safecmd`, `/safepyrun`, `/litesearch`, `/fasthtml`, `/lisette`, `/nbdev`, `/codesigs`
